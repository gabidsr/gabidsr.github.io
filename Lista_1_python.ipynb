```python
#Exercício 5
import math

print("Digite os valores (inteiros e positivos) para descobrir a área do triângulo")
catA = int(input("Cateto A: "))
catB = int(input("Cateto B: "))
hip = int(input("Hipotenusa: "))
s = (catA + catB + hip) / 2
area = math.sqrt(s * (s - catA) * (s - catB) * (s - hip))
print("A área do triângulo é %i" % area)
```

```python
#Exercício 6
print("Digite dois números para descobrir o maior")
num1 = int(input("Primeiro número: "))
num2 = int(input("Segundo número: "))
if num1 > num2:
  print("%i é maior que %i, a diferença entre eles é %i" % (num1, num2, (num1 - num2)))
elif num1 < num2:
  print("%i é maior que %i, a diferença entre eles é %i" % (num2, num1, (num2 - num1)))
else:
  print("%i == %i" % (num1, num2))
```

```python
#Exercício 11
seletor = input("Escolha uma média para calcular:\n"
                "(a) Geométrica\n"
                "(b) Ponderada\n"
                "(c) Harmônica\n"
                "(d) Aritmética\n")

print("Agora digite os valores para calcular a média")
x = int(input("Valor 1: "))
y = int(input("Valor 2: "))
z = int(input("Valor 3: "))

match seletor:
  case 'a':
    media = (x * y * z) ** (1/3)
    print("A média geométrica é %i" % media)
  case 'b':
    media = (x + y * 2 + z * 3) / 6
    print("A média geométrica é %i" % media)
  case 'c':
    media = 1 / ((1 / x) + (1 / y) + (1 / z))
    print("A média geométrica é %i" % media)
  case 'd':
    media = (x + y + z) / 3
    print("A média geométrica é %i" % media)
```

```python
#Exercício 15
print("Calcule quanto deverá pagar ao pedreiro")

alimentacao = float(input("Digite o valor do vale alimentação: "))
transporte = float(input("Digite o valor do transporte unitário: "))
metros = float(input("Digite a quantidade de metros quadrados produzidos: "))

if metros < 10:
  metros_valor = metros * 10.00
elif 10 <= metros < 20:
  metros_valor = metros * 11.50
else:
  metros_valor = metros * 13.00

valor_total = metros_valor + alimentacao + transporte * 2

print("O pedreiro deverá receber R$ $%.2f" % valor_total)
```
